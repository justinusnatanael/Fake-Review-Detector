<p style="text-align:center">Removing URL & HTML Parsing</p>

```python
import pandas as pd
import re
import time

from tqdm import tqdm
```

```python
df = pd.read_csv("../0_SCRAPING/IndoECommerce.csv")
print("Original shape:", df.shape)
df.head()
```

```python
df = df.dropna(subset=["Ulasan"]).copy() # Drop NA
df = df[df["Ulasan"].astype(str).str.strip() != ''] # Remove empty string
df = df.drop_duplicates(subset=["Ulasan"], keep='first') # Remove duplicates
```

```python
url_pattern = re.compile(r'https?://\S+|www\.\S+')
html_pattern = re.compile(r'<.*?>')
underscore_pattern = re.compile(r'_')
space_pattern = re.compile(r'\s+')


def text_cleaner(text):

    text = str(text).lower()

    text = url_pattern.sub('', text)
    text = html_pattern.sub('', text)
    text = underscore_pattern.sub('', text)
    text = space_pattern.sub(' ', text).strip()

    return text
```

```python
df['ulasan_cleaned'] = df['Ulasan'].apply(text_cleaner)
print(f'Raw text:\n{df.Ulasan.iloc[1]}\nClean text:\n{df.ulasan_cleaned.iloc[1]}')
```

```python
print("After clean shape:", df.shape)
df.head()
```

```python
print("📤 Exporting data to CSV...")

pbar = tqdm(total=1, desc="Exporting", leave=True)
df_export = df[['ulasan_cleaned', 'Rating', 'Membantu']]
df_export.to_csv('simple_clean.csv', index=False)
pbar.update(1)
pbar.close()

print(f"✅ Successfully exported {len(df_export)} rows to 'simple_clean.csv'")
```
